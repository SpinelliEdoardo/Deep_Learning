# Perché il linguaggio umano è difficile

## Ambiguità lessicale

- una parola a più significati

## Ambiguita sintattica

- la stessa frase ha più strutture

## Contesto e ironia

- Il senso dipende da chi/qunado/come

## Snonimia e varianti

- Modi diversi e stessi concetti

## Pipeline NLP: dal dato grezzo alla decisione

1. Testo grezzo
    - Recensioni, tweet, email, documenti -----> "IL film non era male"
2. Preprocessing
    - Pulizia TOkenizzazione, normalizzazione
3. Rappresentazione 
    - trasformare le parole in numeri/vettori
4. Modello 
    - ML classico o rete neurale che apprende 
5. Output e valutazion
    - classe, risposte + metriche
