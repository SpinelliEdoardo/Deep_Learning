### One-hot encoding

- per trasformare la parola in vettore (associa ogni aprola ad una posizione del vocabolario, con solo 1 posizione)

#### DIfetti di One-hot encoding:

ecco i due princiapli problemi di questo metodo

- Dimesnione enorme: con 50.000 parole, ogni vettore ha 50.000 dimesnioni (quasi tutte 0), Spreco di memoria

- Nessuna semantica: "gatto" e "gattino" sono ortogonli come "gatto" e "tavolo". La distanza è sempre la stessa: il modello non sa che sono simili

## Bag of Words: contare le parole 

- rappresntiamo un docuemnto con il conteggio di goni parola del vocabolario. L'ordine viene ignorato ( è un "sacco" di parole)

1. Vantaggi: 
- Semplice veloce, sorprendente effcacie per classificare testi

2. Limite:
- Perde l'ordine delle parole; vettori sparsi, tratta tutte le parole come ugualmente importanti, non riconosce sinonimi

## TF-IDF: non tutte le parole valgono uguale

in un insieme di recensioni la parola "film" compare ovunqu: non distingue nulla .
"Capolavoro" invece dice moltissimo

TF-IDF: da pesi alle parile are caratteristiche di un doumento

### Term frequency:

- Quanto questa parola appare in questo documento

(più appare, più è rilevante per il documento)

### inverse document frequency

- Quanto è rara la parola di tutti i docuemnti

(Se appare ovunque (es: "il") -> peso basso. Se è rara -> peso alto)


- fa un prodotto vettroiale tra TF IDF e asocia un peso maggiore, più il termine è frequente più peso da alla mia parola (TF)  

- peso della parola = TF x IDF --> alto solo per parole frequenti nel documento MA rare altrove

### Limiti di Bag of Words e TF-IDF 

questi metodi sono robusti e ancora oggi uati come basline. Ma condividiono devolesse fondamentali

1. Nessun significato

2. Vettori Sparsi: vocabolari da decine di migliori dimesioni, quasi tutte a zero  

3. Ordine perso

4. Fuori vocabolario: Una parola mia vista in addestramento viene semplicemente ignorata

due parole si capisce che sono vicine tramite la similraità del coseno

### Word2Vec: impare i vettori d aun compito finito 

Il modello viene addestrato su un compito specifico **(chiamato pretext task)**. L'output finale della previsione non è ciò che ci interessa. La cosa importante è che, per risolvere questo compito, la macchina è costretta a imparare la relazione tra le parole. In questo modo, parole che compaiono in contesti simili avranno vettori **(embedding)** simili e vicini nello spazio matematico.

- CBOW --> dai il **CONTESTO** e predici la parola

Vantaggio: --> buono con parole frequenti

- SKIP-GRAM ---> dai la **PAROLA**, predici il contesto

# Reti Neurali e sequenze

- perché l'ordine delle parole 

## Reti ricorrenti:

Una RNN elabora la frase parola per parola. A ogni passo aggiorna uno "stato nascosto" che riassume tutto ciò letto fino'ora

### Il problema del vanishing gradient 

le RNN semplici fatiacno con le dipendenze lunghe: il segnale che collega paarole lontane si affievolisce a 0 

- Gradienti che svaniscono: --> le prime parole non ricevono più un sgenale di 
apprendiamento

più va avanti e la rete dimentica quello che sta indietro

- Memoria a breve termine: --> la rete ricorda bene solo le ultime parole

- Difficile parallelizzare: --> deve procedere in sequenza, un passo alla volta lento

## Reti Neurali:

Un neurone combina gli input con dei pesi, somma, e applica funzione non lineare. Impilando neuroni in strati si ottiene una rete capace di apprendere


## LSTM e GRU: memoria che dura ( a lungo termine)

Aggiungono dei "cancelli" (gate) che decidono cosa ricordare, cosa dimenticare e cosa emettere. Così le informazioni sopravvivono a lungo 


- **Forget gate**: --> Decide quali informazioni vecchie buttare via

- **Input gate**: --> Decide quali nuove informazioni memorizzare

- **Output gate**: --> Decide cosa emettere come stato corrente

## Attention

- Per interpretare una parola, la rete "presta attenzione" alle altre parole rilevanti della frase, assegando a ciascuna un peso

la rete guarda tutte le parola all'interno della frase e associa in base alla praola che deve predire un peso di importanza 

### Self-Attention 
- ogni parola genera tre vettori. Un'analogia con una libreria

1. Query ---> Cosa sto cercando?
la domanda che la parola pone alle altre 

2. Key --> Cosa offro
L'etichetta con cui ogni parola si fa trovare

3. Value --> Il contenuto
l'informazione effettiva da raccogliere

**Il Meccanismo, in una frase**

-la Query di una parola viene confrontata con la key di tutte le altre (prodotto scalare) -> si ottengono i pesi di attenzione -> si combinano i Value pesanti. Risultato: ogni parola diventa una miscela contesutuale di altre
