## 1. Tokenizzazione e conteggio dei token

```python
import re
import nltk
import string 
import spacy
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem.snowball import ItalianStemmer

# scarichiamo il pacchetto per la tokenizzazione se non presente
nltk.download('punkt')
nltk.download('punkt_tab')

recensioni = [
    # Prodotto fisico / Tech
    "Spedizione velocissima! 📦🚚 Pacco arrivato intatto ✨... Le cuffie suonano bene 🎧, ma il cavo è davvero TROPPO corto!! 😡 Voto: 4/5 ★★★★☆",
    "Pessimo! ❌ Dò 1/5 ⭐️. Dopo appena 2 giorni (48h!) ha smesso di funzionare 💥. Ho avviato il reso... #Deluso 💸👎",
    "Rapporto qualità/prezzo TOP!! 🔥⚡️ Materiali 100% resistenti e design curato 👌. Consigliatissimo @tutti 💯💯",
    "Servizio ok 🤷‍♂️, ma la qualità *non* rispecchia le foto... 📉 Piuttosto deluso ~_~ (prezzo alto vs qualità mediocre) 💰➡️🗑️",

    # Ristorante / Food
    "Cibo ECCELLENTE!! 🍕🍷 Ingredienti 100% freschi 🌿. Personale super cortese ❤️, anche se ~40 min di attesa... ⏳️",
    "Pizza ottima 😋 e ben lievitata 👏! Però locale rumoroso (troppo BASS!!) 📢 e tavoli appiccicati... 🛋️=🛋️ Voto 3.5/5 ★★★☆☆",

    # Servizi / Hotel
    "Soggiorno PER-FET-TO! 🏨✨ Camera pulitissima 🧹& colazione SUPER abbondante 🥐☕️. Torneremo sicuramente!! ✈️🇮🇹",
    "Assistenza clienti pessima 📧❌... Lenta & poco risolutiva! Ho chiamato x3 volte! ☎️💥 #PessimoServizio"
]

# Uniamo la lista di stringhe in un'unica stringa
testo_unito = " ".join(recensioni)
risultato = re.findall(r'pattern', testo_unito)

# tokenizzazione 

token = word_tokenize(testo_unito, language='italian')

# Conteggio

token_totali = len(token)
token_unici = len(set(token))

print(f"token totali: {token_totali}")
print(f"Token unici: {token_unici}")
```

## 2. Lowercasing, rimozione punteggiatura e stopwords

- Portiamo il testo in minuscolo e puliamo i token irrilevanti.

```python
import string 
from nltk.corpus import stopwords

nltk.download('stopwords')
stop_words = set(stopwords.words('italian'))

# lowercasing, e filtraggio di punteggiatura/emoji/stopwords

token_puliti = [
    t.lower() for t in token 
    if t.isalnum() and t.lower not in stop_words
]

print("Token puliti", token_puliti)
```

# 3. Confronto tra Stemming e Lemmatization

```python
import spacy
from nltk.stem.snowball import ItalianStemmer

# Stemming con NLTK

stemmer = ItalianStemmer()
stemmed = [stemmer.stem(t) for t in token_puliti]

# Lemmatization con spaCy (modello italiano)
# Per installare il modello: !python -m spacy download it_core_news_sm

nlp = spacy.load("it_core_news_sm")
doc = nlp(" ".join(token_puliti))
lemmatized = [token.lemma_ for token in doc]

# confronto visivo

for orig, st, lem in zip(token_puliti, stemmed, lemmatized):
    print(f"Originale: {orig:<12} | Stem: {st:<12} | Lemma: {lem}")
```

## 4. Estrazione di hashtag e cifre con Regex

```python
import re

# Estrazione della cifre (es: 4 e 5)
cifre = re.findall(r'\d+', testo_unito)

# Estrazione degli hashtag (es: #esempio)
hashtag = re.findall(r'#\w+', testo_unito)

print("Cifre trovate:", cifre)
print("Hashtag trovati:", hashtag) # Risulterà vuota poiché non ce ne sono nel testo
```
