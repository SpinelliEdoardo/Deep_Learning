## scrivi il nome si una funzione che dato il testo di un sistema restituisce la stringa "AI, ML o DL"

regole da applicare:

1. se il sistema segue regole scritte a mano -> AI
2. Se impara dai dati da un algoritmo calssico -> ML
3. Se usa una rete neurale profonda -> DL

## verifica la tua funzione sui casi di una lista 

creare una lista di esempi:

es:

casi: [] associare i nomi delle stringhe 

fare ciclo che prende una lista in ingresso e ti da la ripsota della categoria

```python
def classifica_sistema(descrizione_sistema):
    """
    Classifica un sistema in AI, ML o DL in base al suo funzionamento.
    """
    descrizione = descrizione_sistema.lower()
    
    if "rete neurale profonda" in descrizione or "deep learning" in descrizione:
        return "DL"
    elif "machine learning" in descrizione or "algoritmo classico" in descrizione or "addestrato" in descrizione:
        return "ML"
    elif "artificial intelligence" in descrizione or "regole scritte a mano" in descrizione or "istruzioni" in descrizione:
        return "AI"
    else:
        return "Non Classificabile"
```

```python
casi_di_testo = [
    "Un programma che valuta le richieste di prestito usando regole scritte a mano dagli esperti.",
    "Un modello di classificazione spam che impara dai dati ricevuti tramite un algoritmo classico come la regressione logistica.",
    "Un sistema di riconoscimento facciale che usa una rete neurale profonda a molti strati.",
    "Un chatbot basato su un insieme di istruzioni IF-THEN definite manualmente.",
    "Un software che predice il prezzo delle case addestrato su vendite passate.",
    "ciao"
]
```

```python
for i, testo in enumerate(casi_di_testo, 1):
    categoria = classifica_sistema(testo)
    print(f"Caso {i}: {testo}")
    print(f" Categoria: {categoria}\n")
```
